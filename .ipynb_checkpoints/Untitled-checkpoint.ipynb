```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Setup Data
np.random.seed(42)
n_points = 20_000

# Generate points in a unit disk
r = np.sqrt(np.random.rand(n_points))
theta = np.random.rand(n_points) * 2 * np.pi
x = r * np.cos(theta)
y = r * np.sin(theta)
X = np.column_stack((x, y))

# Parameters
w = np.array([1.5, 0.5])  # Weight vector
b = 0.5                   # Bias
norm_w = np.linalg.norm(w)

# 2. Computations
# The scalar field (dot product) BEFORE bias
dot_products = X.dot(w)

# The Linear Step (z = w*x + b)
z = dot_products + b

# The Nonlinear Step (activation)
activation = np.tanh(z)

# 3. Plotting
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# --- Plot 1: Input Space (The Scalar Field) ---
# Color points by their projection value to track them through the transforms
sc1 = axes[0].scatter(x, y, c=dot_products, cmap='viridis', s=10, alpha=0.6)
axes[0].arrow(0, 0, w[0], w[1], head_width=0.1, head_length=0.1, fc='red', ec='red', linewidth=2, label='Vector w')
axes[0].set_title("1. Input Space (2D Disk)\nColor = alignment with w")
axes[0].set_xlabel("x1")
axes[0].set_ylabel("x2")
axes[0].axis('equal')
axes[0].grid(True, linestyle='--', alpha=0.5)
axes[0].legend()

# --- Plot 2: The Linear Collapse (Projection onto w + bias) ---
# We plot the values on the x-axis, and random jitter on y just to see the density
jitter = np.random.normal(0, 0.05, n_points)
axes[1].scatter(z, jitter, c=dot_products, cmap='viridis', s=10, alpha=0.6)
axes[1].set_title(f"2. Linear Collapse (1D Line)\nz = w·x + {b}")
axes[1].set_xlabel("z value")
axes[1].set_yticks([]) # Hide y axis as it has no meaning here
axes[1].grid(True, axis='x', linestyle='--', alpha=0.5)
# Mark the theoretical bounds [-||w||+b, ||w||+b]
min_bound = -norm_w + b
max_bound = norm_w + b
axes[1].axvline(min_bound, color='k', linestyle=':', label='Min/Max Reach')
axes[1].axvline(max_bound, color='k', linestyle=':')
axes[1].text(min_bound, 0.2, f"{min_bound:.2f}", ha='center')
axes[1].text(max_bound, 0.2, f"{max_bound:.2f}", ha='center')

# --- Plot 3: The Sigmoid Warp ---
# Plot z vs tanh(z)
axes[2].scatter(z, activation, c=dot_products, cmap='viridis', s=10, alpha=0.6)
axes[2].set_title("3. Sigmoid Deformation\ny = tanh(z)")
axes[2].set_xlabel("z (Linear Output)")
axes[2].set_ylabel("Activation (tanh)")
axes[2].grid(True, linestyle='--', alpha=0.5)
# Draw the actual tanh curve for reference
z_space = np.linspace(z.min(), z.max(), 100)
axes[2].plot(z_space, np.tanh(z_space), 'k--', alpha=0.5, label='tanh(z)')

plt.tight_layout()
plt.show()
```
